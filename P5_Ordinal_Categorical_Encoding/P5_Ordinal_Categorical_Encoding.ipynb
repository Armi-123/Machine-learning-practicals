# Practical 5: Ordinal Categorical Encoding Using Pandas Replace

## Aim
To transform ordinal categorical string labels into numerical equivalents using the Pandas DataFrame `replace()` method.

## Objectives
- Create a DataFrame containing ordinal categorical data.
- Understand ordinal categorical features.
- Convert string labels into numerical values.
- Use Pandas `replace()` method for ordinal encoding.
- Display the original and encoded data.

```python
# Import Pandas library
import pandas as pd
```

## 1. Create Dataset

We create a sample dataset containing an ordinal categorical feature called `Performance`.

The categories are:

- Low
- Medium
- High

These categories have a meaningful order:

**Low < Medium < High**

```python
# Create sample student data
data = {
    "Student": ["Armi", "Rahul", "Priya", "Amit", "Neha", "Riya"],
    "Performance": [
        "High",
        "Medium",
        "Low",
        "High",
        "Medium",
        "Low"
    ]
}

# Create a Pandas DataFrame
df = pd.DataFrame(data)

# Display the original DataFrame
print("Original Data:")
print(df)
```

## 2. Ordinal Categorical Feature

`Performance` is an ordinal categorical feature because its categories have a natural ranking.

The order is:

Low < Medium < High

We convert these categories into numerical values:

- Low → 1
- Medium → 2
- High → 3

```python
# Display unique categories in the Performance column
print("Unique Categories:")
print(df["Performance"].unique())
```

## 3. Convert String Labels to Numerical Values

The Pandas `replace()` method is used to replace the string labels with numerical values.

Mapping:

- Low → 1
- Medium → 2
- High → 3

```python
# Create a mapping for ordinal categories
ordinal_mapping = {
    "Low": 1,
    "Medium": 2,
    "High": 3
}

# Replace categorical labels with numerical values
df["Performance"] = df["Performance"].replace(ordinal_mapping)

# Display the encoded DataFrame
print("Encoded Data:")
print(df)
```

## 4. Compare Original and Encoded Data

The original categorical labels have now been replaced with numerical values.

For example:

- High becomes 3
- Medium becomes 2
- Low becomes 1

```python
# Display the final encoded DataFrame
print("========== FINAL ENCODED DATA ==========")
print(df)

# Display the data types
print("\n========== DATA TYPES ==========")
print(df.dtypes)
```
