# Practical 2: Filling Missing Values Using K-Nearest Neighbors (KNN)

## Aim
To use the K-Nearest Neighbors (KNN) algorithm to fill missing values in a dataset.

## Objectives
- Create a dataset containing missing values.
- Identify missing values in the dataset.
- Use KNNImputer to fill the missing values.
- Display the dataset before and after imputation.

```python
# Import Pandas for data manipulation
import pandas as pd

# Import KNNImputer from Scikit-learn
from sklearn.impute import KNNImputer
```

## 1. Create Dataset

First, we create a sample dataset containing student information.

Some values in the `Age`, `Marks`, and `Attendance` columns are intentionally set as missing values (`NaN`).

```python
# Create sample student data
data = {
    "Age": [21, 22, 20, None, 23, 21, None],
    "Marks": [85, 78, None, 88, 92, None, 80],
    "Attendance": [90, 85, 88, None, 95, 92, 80]
}

# Create a DataFrame
df = pd.DataFrame(data)

# Display the original dataset
print("Original Dataset:")
print(df)
```

## 2. Check Missing Values

Before applying KNN, we check how many missing values are present in each column.

```python
# Count missing values in each column
missing_values = df.isnull().sum()

# Display the number of missing values
print("Missing Values:")
print(missing_values)
```

## 3. Apply KNN Imputation

`KNNImputer` fills missing values by finding the nearest rows based on the available values.

The `n_neighbors` parameter determines how many neighboring rows are considered for calculating the missing value.

Here, we use `n_neighbors = 3`.

```python
# Create a KNN Imputer
# n_neighbors=3 means the 3 nearest rows will be used
# to estimate the missing values
imputer = KNNImputer(n_neighbors=3)

# Apply KNN imputation to the dataset
df_imputed = imputer.fit_transform(df)

# Convert the resulting NumPy array back into a DataFrame
df_imputed = pd.DataFrame(df_imputed, columns=df.columns)

# Display the dataset after imputation
print("Dataset After KNN Imputation:")
print(df_imputed)
```

## 4. Compare Original and Imputed Data

Now we compare the original dataset containing missing values with the dataset after applying KNN imputation.

```python
# Display original dataset
print("========== ORIGINAL DATA ==========")
print(df)

# Display imputed dataset
print("\n========== DATA AFTER KNN IMPUTATION ==========")
print(df_imputed)
```

## 5. Verify Missing Values

Finally, we check the dataset again to confirm that the missing values have been filled.

```python
# Check for missing values after KNN imputation
remaining_missing = df_imputed.isnull().sum()

print("Missing values after KNN imputation:")
print(remaining_missing)
```

## Result

The K-Nearest Neighbors (KNN) imputation technique was successfully used to fill the missing values in the dataset.

The `KNNImputer` identified neighboring observations and estimated the missing values based on those observations. After imputation, the dataset contained no missing values.
