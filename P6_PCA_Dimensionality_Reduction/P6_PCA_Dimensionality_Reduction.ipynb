# Practical 6: Dimensionality Reduction Using PCA

## Aim
To reduce the dimensions of a dataset using Principal Component Analysis (PCA).

## Objectives
- Create a dataset with multiple features.
- Standardize the features.
- Apply Principal Component Analysis (PCA).
- Reduce the number of dimensions.
- Analyze the explained variance.
- Display the transformed dataset.

```python
# Import Pandas for data handling
import pandas as pd

# Import StandardScaler for feature standardization
from sklearn.preprocessing import StandardScaler

# Import PCA for dimensionality reduction
from sklearn.decomposition import PCA
```

## 1. Create Dataset

We create a dataset containing four numerical features.

The four features are:

- Study Hours
- Attendance
- Assignment Score
- Previous Marks

PCA will be used to reduce these four dimensions into two principal components.


```python
# Create sample student dataset
data = {
    "Study_Hours": [2, 4, 6, 8, 10, 3, 5, 7, 9, 11],
    "Attendance": [60, 65, 70, 80, 90, 62, 68, 75, 85, 92],
    "Assignment_Score": [50, 55, 65, 72, 88, 52, 60, 70, 80, 90],
    "Previous_Marks": [45, 50, 62, 70, 85, 48, 58, 68, 78, 88]
}

# Create a Pandas DataFrame
df = pd.DataFrame(data)

# Display the original dataset
print("Original Dataset:")
print(df)
```

## 2. Select Features

We select the numerical features that will be used for PCA.

The original dataset contains four dimensions/features.

```python
# Select all numerical features for PCA
X = df[
    [
        "Study_Hours",
        "Attendance",
        "Assignment_Score",
        "Previous_Marks"
    ]
]

# Display the features
print("Features:")
print(X)

# Display the original number of dimensions
print("\nOriginal Number of Dimensions:", X.shape[1])
```

## 3. Standardize the Features

PCA is affected by the scale of the features.

Therefore, we standardize the features so that they have a similar scale before applying PCA.

We use `StandardScaler()` for standardization.

```python
# Create StandardScaler object
scaler = StandardScaler()

# Standardize the features
X_scaled = scaler.fit_transform(X)

# Display the standardized data
print("Standardized Data:")
print(X_scaled)
```

## 4. Apply Principal Component Analysis

We apply PCA and reduce the original four dimensions into two principal components.

Therefore:

**4 dimensions → 2 dimensions**

```python
# Create PCA object to reduce the data to 2 components
pca = PCA(n_components=2)

# Transform the standardized data using PCA
X_pca = pca.fit_transform(X_scaled)

# Convert PCA result into a DataFrame
pca_df = pd.DataFrame(
    X_pca,
    columns=["Principal_Component_1", "Principal_Component_2"]
)

# Display the reduced dataset
print("PCA Reduced Dataset:")
print(pca_df)
```

## 5. Explained Variance

Explained variance tells us how much information from the original dataset is retained by each principal component.

We calculate the explained variance ratio of the two principal components.

```python
# Get the explained variance ratio
explained_variance = pca.explained_variance_ratio_

# Display explained variance
print("Explained Variance Ratio:")
print(explained_variance)

# Calculate total variance retained
total_variance = explained_variance.sum()

print("\nTotal Variance Retained:",
      total_variance * 100, "%")
```

## 6. Compare Original and Reduced Dimensions

Before PCA, the dataset had four features.

After PCA, the dataset contains only two principal components.

Thus, PCA reduced the dimensionality from four dimensions to two dimensions.

```python
# Display original dimensions
print("Original Dimensions:", X.shape[1])

# Display reduced dimensions
print("Reduced Dimensions:", X_pca.shape[1])

# Display reduction
print(
    "\nDimensions reduced from",
    X.shape[1],
    "to",
    X_pca.shape[1]
)
```

## Result

Principal Component Analysis (PCA) was successfully implemented to reduce the dimensionality of the dataset.

The original dataset contained four features, which were reduced to two principal components.

The explained variance ratio was calculated to determine how much information was retained after dimensionality reduction.
