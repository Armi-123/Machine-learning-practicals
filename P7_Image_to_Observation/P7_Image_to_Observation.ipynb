# Practical 7: Converting an Image into an Observation for Machine Learning

## Aim
To convert an image into a numerical observation that can be used as input for a machine learning algorithm.

## Objectives
- Create a simple image.
- Read the image using Python.
- Convert the image into pixel values.
- Flatten the pixel values into a one-dimensional observation.
- Understand how images are represented as numerical data.

```python
# Import NumPy for numerical array operations
import numpy as np

# Import Matplotlib for creating and displaying the image
import matplotlib.pyplot as plt
```

## 1. Create a Simple Image

A grayscale image can be represented as a 2D matrix.

Each value represents the intensity of a pixel:

- 0 → Black
- 255 → White
- Values between 0 and 255 → Different shades of gray

We create a simple 5 × 5 grayscale image.

```python
# Create a 5 × 5 grayscale image
image = np.array([
    [0,   0,   0,   0,   0],
    [0, 255, 255, 255, 0],
    [0, 255,   0, 255, 0],
    [0, 255, 255, 255, 0],
    [0,   0,   0,   0,   0]
])

# Display the image
plt.imshow(image, cmap="gray")
plt.title("Sample Grayscale Image")
plt.axis("off")
plt.show()
```

## 2. Check Image Dimensions

The image has 5 rows and 5 columns.

Therefore, it contains:

5 × 5 = 25 pixels.

```python
# Display image dimensions
print("Image Shape:", image.shape)

# Calculate total number of pixels
print("Total Pixels:", image.size)
```

## 3. Convert Image into a Numerical Observation

Machine learning algorithms generally work with numerical feature vectors.

Therefore, we flatten the 2D image matrix into a 1D array.

A 5 × 5 image containing 25 pixels becomes one observation containing 25 numerical features.

```python
# Flatten the 2D image into a 1D array
observation = image.flatten()

# Display the numerical observation
print("Numerical Observation:")
print(observation)

# Display number of features
print("\nNumber of Features:", len(observation))
```

## 4. Normalize Pixel Values

Pixel values in a grayscale image usually range from 0 to 255.

For machine learning, we can normalize these values to a range between 0 and 1.

Formula:

Normalized Pixel = Pixel / 255

```python
# Normalize pixel values from 0-255 to 0-1
normalized_observation = observation / 255.0

# Display normalized observation
print("Normalized Observation:")
print(normalized_observation)
```

## 5. Convert Observation into Machine Learning Input

Machine learning models generally expect the input in the form:

(number of observations, number of features)

Here we have:

- 1 observation
- 25 features

Therefore, the input shape will be `(1, 25)`.

```python
# Reshape the observation for machine learning
X = normalized_observation.reshape(1, -1)

# Display the machine learning input
print("Machine Learning Input:")
print(X)

# Display shape
print("\nInput Shape:", X.shape)
```

## Result

The image was successfully converted into a numerical observation for machine learning.

The 5 × 5 grayscale image contained 25 pixels. These pixel values were flattened into a one-dimensional feature vector and normalized between 0 and 1.

Thus, the image was converted into a machine-learning input with 1 observation and 25 numerical features.
