# Practical 1: Read and Write CSV and Excel Files

## Aim
To write a Python program using Pandas to read and write data in CSV and Excel files.

## Objectives
- Create a DataFrame using Pandas.
- Write DataFrame data into a CSV file.
- Read data from a CSV file.
- Write DataFrame data into an Excel file.
- Read data from an Excel file.

```python
# Import Pandas library
import pandas as pd
```

## 1. Create Sample Data

First, we create sample student data using a Python dictionary and convert it into a Pandas DataFrame.

```python
# Create sample student data using a dictionary
data = {
    "Name": ["Armi", "Rahul", "Priya", "Amit", "Neha"],
    "Age": [21, 22, 21, 23, 22],
    "Marks": [85, 78, 92, 88, 95]
}

# Convert the dictionary into a Pandas DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
print("Original Data:")
print(df)
```

## 2. Write Data to CSV File

The `to_csv()` function is used to save a Pandas DataFrame into a CSV file.

The `index=False` parameter prevents the DataFrame index from being saved as an extra column.

```python
# Write the DataFrame to a CSV file
df.to_csv("students.csv", index=False)

# Display confirmation message
print("Data successfully written to students.csv")
```

```python
excel_data = pd.read_excel("students.xlsx")
print(excel_data)
```

## 3. Read Data from CSV File

The `read_csv()` function is used to read data from a CSV file and store it in a Pandas DataFrame.

```python
# Read the CSV file using Pandas
csv_data = pd.read_csv("students.csv")

# Display the data read from the CSV file
print("Data read from CSV file:")
print(csv_data)
```

## 4. Write Data to Excel File

The `to_excel()` function is used to save a Pandas DataFrame into an Excel file.

The Excel file is saved with the `.xlsx` extension.

```python
# Write the DataFrame to an Excel file
df.to_excel("students.xlsx", index=False)

# Display confirmation message
print("Data successfully written to students.xlsx")
```

## 5. Read Data from Excel File

The `read_excel()` function is used to read data from an Excel file and store it in a Pandas DataFrame.

```python
# Read the Excel file using Pandas
excel_data = pd.read_excel("students.xlsx")

# Display the data read from the Excel file
print("Data read from Excel file:")
print(excel_data)
```

## 6. Display Final Results

Finally, we display the data obtained from both the CSV and Excel files to verify that the read and write operations were successful.

```python
# Display data read from CSV file
print("========== CSV DATA ==========")
print(csv_data)

# Display data read from Excel file
print("\n========== EXCEL DATA ==========")
print(excel_data)

# Display completion message
print("\nPractical 1 completed successfully!")
```

## Result

The Python program successfully performed the following operations:

1. Created a Pandas DataFrame.
2. Wrote the DataFrame into a CSV file.
3. Read data from the CSV file.
4. Wrote the DataFrame into an Excel file.
5. Read data from the Excel file.

Thus, the read and write operations for CSV and Excel files were successfully implemented using Python and Pandas.
